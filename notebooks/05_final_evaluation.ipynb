# Final Evaluation of ASL Fingerspelling Classifier

In this notebook, we will evaluate the performance of the best model obtained from hyperparameter tuning on the validation dataset. We will generate predictions, visualize the results, and analyze the model's performance.

```python
# Import necessary libraries
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
from asl_classifier.utils.config import load_config, BASE_DIR, MODEL_CONFIG_PATH


# Load configuration
config = load_config(MODEL_CONFIG_PATH)  # Load model and training configurations
print(config)

# Load the best model
model_path = os.path.join(BASE_DIR, config['model_path'])
best_model = tf.keras.models.load_model(model_path)  # Load the best model path from config
print("Best model loaded successfully.")
```

```python
# Load validation dataset
val_ds = tf.keras.preprocessing.image_dataset_from_directory(
    config['validation_data_path'],  # Path to validation data
    image_size=(200, 200),
    batch_size=32
)
print(f"Validation dataset loaded with {len(val_ds)} batches.")
```

```python
# Evaluate model performance
val_loss, val_accuracy = best_model.evaluate(val_ds, verbose=0)
print("Validation Loss: {:.4f}".format(val_loss))
print("Validation Accuracy: {:.4f} ({:.2f}%)".format(val_accuracy, val_accuracy * 100))
```

```python
# Generate predictions for validation set
y_true = []
y_pred = []

for images, labels in val_ds:
    preds = best_model.predict(images, verbose=0)
    y_true.extend(labels.numpy())
    y_pred.extend(np.argmax(preds, axis=1))
```

```python
# Classification report
print("\nDetailed Classification Report:")
print("=" * 80)
print(classification_report(y_true, y_pred, target_names=val_ds.class_names))
```

```python
# Confusion matrix visualization
cm = confusion_matrix(y_true, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=val_ds.class_names)

fig, ax = plt.subplots(figsize=(14, 14))
disp.plot(ax=ax, xticks_rotation=45, cmap='Blues', values_format='d')
plt.title("Confusion Matrix - ASL Alphabet Classifier", fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```
